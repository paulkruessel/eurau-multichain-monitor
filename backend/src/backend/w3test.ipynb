# Web3 Package test
## Goal: Connect to allUnions EURAU RPC API

Import web3

```python
from web3 import Web3, AsyncWeb3
```

```python
# read out address and contract address from .env

from dotenv import load_dotenv
from pathlib import Path
import os

env_path = Path("../../.env").resolve()

print(env_path)
print(env_path.exists())

load_dotenv(env_path)

rpc_url = os.environ["BASE_RPC_URL"]
contract_address = os.environ["EURAU_BASE_CONTRACT_ADDRESS"]

print(rpc_url)
print(contract_address)
```

Get HTTP Provider

```python
w3 = Web3(Web3.HTTPProvider(rpc_url))
w3.is_connected()
```
